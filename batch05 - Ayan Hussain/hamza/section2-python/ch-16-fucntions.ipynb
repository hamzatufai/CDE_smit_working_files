```python
def apply_text(amount, rate=0.15):
    return amount + (amount * rate)

apply_text(1000)
apply_text(2000, 0.10)
```

```python
def apply_text(amount, rate=0.15, discount=0.10):
    return amount + (amount * rate) - (amount * discount)


apply_text(1000)
```

```python
# add, sub, multiply, divide
def calculator(num1, num2):
    print("=/============ Calculator =============/=")
    print("1) add, 2) sub, 3) multiply, 4) divide")
    user_choice = int(input("Select Operator e.g.(1, 2): "))
    if user_choice == 1:
        return num1 + num2
    elif user_choice == 2:
        return num1 - num2
    elif user_choice == 3:
        return (num1 * num2)
    elif user_choice == 4:
        return num1 / num2
    else:
        print("Invalid user input. ")

calculator(2, 7)
```
