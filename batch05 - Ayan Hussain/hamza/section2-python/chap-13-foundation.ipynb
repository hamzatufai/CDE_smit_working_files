```python
# Equality Operator ==
value_1 = 5
value_2 = 5

print(value_1 == value_2)
print(value_1 != value_2)

print(value_1 < value_2)  # < less than  --- false
print(value_1 <= value_2)  # < less than equal --- true

print(value_1 > value_2)  # > greater than  ------false
print(value_1 >= value_2)  # > greater than equal ------- True
```

```python
# Equality Operator
value_1 = 5
value_2 = 6

print(value_1 == value_2)
print(value_1 != value_2)

print(value_1 < value_2)  # < less than
print(value_1 <= value_2)  # < less than equal

print(value_1 > value_2)  # > greater than
print(value_1 >= value_2)  # > greater than equal
```
